```python
import networkx as nx
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
```

```python
n = 500
m1 = 100
m2 = 300
m3 = 450

G1 = nx.barabasi_albert_graph(n, m1)
G2 = nx.barabasi_albert_graph(n, m2)
G3 = nx.barabasi_albert_graph(n, m3)

graphs = [(G1, m1), (G2, m2), (G3, m3)]
print(graphs[0][0])
print(graphs[1][0])
print(graphs[2][0])
```

```python
# Create plot
plt.figure(figsize=(15, 4))

for i, (G, m) in enumerate(graphs, 1):
    degrees = [d for _, d in G.degree()]
    max_deg = max(degrees)

    degree_counts = [degrees.count(k) for k in range(max_deg + 1)]

    plt.subplot(1, 3, i)
    plt.loglog(range(max_deg + 1), degree_counts,
               marker='o', linestyle='none')
    plt.title(f"Degree distribution (m={m})")
    plt.xlabel("Degree (k)")
    plt.ylabel("Number of nodes")
    plt.grid(True, which="both", ls="--", alpha=0.4)

plt.tight_layout()
plt.show()
```

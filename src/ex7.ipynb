```python
import networkx as nx
import random
```

```python
def percolate(G, p):
    H = nx.DiGraph()
    H.add_nodes_from(G.nodes())
    for u, v in G.edges():
        if random.random() < p:
            H.add_edge(u, v)
    return H

def spread(H, A):
    seen = set(A)
    stack = list(A)
    while stack:
        x = stack.pop()
        for y in H.successors(x):
            if y not in seen:
                seen.add(y)
                stack.append(y)
    return len(seen)
```

```python
def test_ex5(G, p=0.5):
    H = percolate(G, p)

    nodes = list(G.nodes())
    random.shuffle(nodes)

    A1 = {nodes[0]}
    A2 = {nodes[0], nodes[1]}
    u  = nodes[2]

    lhs = spread(H, A1 | {u}) - spread(H, A1)
    rhs = spread(H, A2 | {u}) - spread(H, A2)

    print("A1:", A1)
    print("A2:", A2)
    print("u:", u)
    print("Added infections A1 → A1∪{u}:", lhs)
    print("Added infections A2 → A2∪{u}:", rhs)
    print("result:", lhs >= rhs)
    print()
```

```python
G1 = nx.path_graph(6, create_using=nx.DiGraph)
G2 = nx.cycle_graph(6, create_using=nx.DiGraph) 
G3 = nx.star_graph(5)       
G3 = nx.DiGraph(G3)                              
```

```python
test_ex5(G1)
test_ex5(G2)
test_ex5(G3)
```

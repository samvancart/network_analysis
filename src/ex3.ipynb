```python
import networkx as nx
import numpy as np
import matplotlib.pyplot as plt
```

```python
n = 600                               
p_values = np.linspace(0.001, 0.05, 10) 
num_instances = 10
```

```python
connectivity_freq = []
largest_cc_sizes = []

for p in p_values:
    connected_count = 0
    cc_sizes = []
    
    for _ in range(num_instances):
        G = nx.gnp_random_graph(n, p)
        
        # Connectivity
        if nx.is_connected(G):
            connected_count += 1
        
        # Largest connected component
        cc = max(nx.connected_components(G), key=len)
        cc_sizes.append(len(cc))
    
    connectivity_freq.append(connected_count / num_instances)
    largest_cc_sizes.append(np.mean(cc_sizes))
```

```python
plt.figure(figsize=(7,5))
plt.scatter(p_values, connectivity_freq)

plt.title("Plot 01: Connectivity Frequency vs p\n"
          "Relative frequency of connected G(n,p) graphs")
plt.xlabel("p")
plt.ylabel("Connectivity frequency")
plt.grid(alpha=0.3)

# Personal path ../data/TutorialC/plots/
plt.savefig("plot_01_relative_frequency.png", dpi=300, bbox_inches="tight")
plt.close()
```

```python
plt.figure(figsize=(7,5))
plt.scatter(p_values, largest_cc_sizes)

plt.title("Plot 02: Largest Component Size vs p\n"
          "Average size of the largest connected component in G(n,p)")
plt.xlabel("p")
plt.ylabel("Average largest component size")
plt.grid(alpha=0.3)

# Personal path ../data/TutorialC/plots/
plt.savefig("plot_02_average_size.png", dpi=300, bbox_inches="tight")
plt.close()
```
